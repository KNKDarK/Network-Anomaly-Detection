## Load data
Use the shared preprocessing utilities.

```python
import sys
sys.path.insert(0, '..')

from src.data_utils import load_raw_csvs, clean_features

raw = load_raw_csvs()
X, y = clean_features(raw)
X.head()
```

## Class distribution
Normal vs Attack imbalance.

```python
import matplotlib.pyplot as plt
import seaborn as sns

y.value_counts().plot(kind='bar', figsize=(10,4))
plt.title('Class distribution')
plt.show()
```

## Missing values & data types

```python
print('Nulls:')
print(X.isnull().sum().sum())
print('\nData types:')
print(X.dtypes.value_counts())
```

## Correlation heatmap
Top correlated numeric features.

```python
corr = X.select_dtypes('number').corr().abs()
plt.figure(figsize=(10,8))
sns.heatmap(corr, cmap='viridis')
plt.title('Correlation heatmap')
plt.show()
```
